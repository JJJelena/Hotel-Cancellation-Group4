# A01 - Data Inspection

Hotel Booking Cancellation Prediction

## Purpose
Inspect the raw dataset before cleaning and modeling.

## Checks
- Dataset structure
- Missing values
- Duplicate records
- Outliers
- Cancellation distribution

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Python is ready!")
print("Pandas version:", pd.__version__)
```

```python
from pathlib import Path
import os

# Build the path from this notebook's location, so it works on teammates' computers too.
notebook_path = os.environ.get("JPY_SESSION_NAME")
if notebook_path:
    project_root = Path(notebook_path).resolve().parent.parent
else:
    project_root = Path.cwd().resolve()
    if project_root.name == "notebooks":
        project_root = project_root.parent

csv_path = project_root / "data" / "hotel_bookings.csv"
df = pd.read_csv(csv_path)

print("CSV path:", csv_path)
print("Dataset shape:", df.shape)
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.columns.tolist()
```

```python
df.info()
```

```python
missing_table = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percent": (df.isna().mean() * 100).round(2)
})

missing_table = missing_table.sort_values(
    by="Missing_Percent",
    ascending=False
)

missing_table
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

print(
    "Duplicate percentage:",
    round(duplicate_count / len(df) * 100, 2),
    "%"
)
```

```python
df.describe().T
```

```python
print("Negative ADR:", (df["adr"] < 0).sum())

print("Zero adults:", (df["adults"] == 0).sum())

total_guests = (
    df["adults"]
    + df["children"].fillna(0)
    + df["babies"]
)

print("Zero total guests:", (total_guests == 0).sum())

total_nights = (
    df["stays_in_week_nights"]
    + df["stays_in_weekend_nights"]
)

print("Zero total nights:", (total_nights == 0).sum())
```

```python
print("Cancellation counts:")
print(df["is_canceled"].value_counts())

print("\nCancellation rates (%):")
print(
    (df["is_canceled"].value_counts(normalize=True) * 100).round(2)
)
```

```python
output_dir = project_root / "outputs"

output_dir.mkdir(parents=True, exist_ok=True)

missing_table.to_csv(
    output_dir / "missing_values_report.csv",
    index=True
)

print("File saved to:", output_dir)
```

## Inspection Summary

- The dataset contains 119,390 rows and 32 columns.
- The overall cancellation rate is 37.04%.
- Four columns contain missing values.
- There are 31,994 fully duplicated rows, which will not be automatically removed.
- Potential anomalies include negative ADR, zero guests, and zero-night stays.
- Further cleaning and leakage prevention are required before modeling.
