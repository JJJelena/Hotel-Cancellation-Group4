## Threshold and sigmoid protocol

Raw thresholds and sigmoid thresholds maximise selected-model training OOF F1 on 0.05–0.95 in 0.05 steps. Sigmoid is cross-fitted over existing OOF folds for the validation grid, then fitted to all OOF pairs for test transformation. Test labels never fit a map or choose a threshold. This practical assessment is post-selection rather than fully nested.
