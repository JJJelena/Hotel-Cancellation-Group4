# A 04 Common preprocessing

The final supervised contract has 24 inputs. Preprocessing is fitted inside each model training fold. This notebook validates the source-built template without replacing the supplied model files.

```python
from pathlib import Path
import sys
project_root = Path.cwd().resolve()
if project_root.name == 'notebooks':
    project_root = project_root.parent
assert (project_root / 'data/hotel_bookings_clean.csv').is_file(), 'Run from the project root or notebooks directory.'
sys.path.insert(0, str(project_root / 'src'))
from preprocessing_source import load_shared_data
from sklearn.base import clone
data = load_shared_data(project_root)
print("Inputs:",len(data["features"]))
print(data["features"])
```

```python
preprocessor = clone(data['preprocessor'])
train, test = data['random_train'], data['random_test']
X_train = preprocessor.fit_transform(train[data['features']])
X_test = preprocessor.transform(test[data['features']])
assert X_train.shape == (94851, 221)
assert X_test.shape == (23713, 221)
assert not hasattr(data['preprocessor'], 'transformers_')
print('Training matrix:', X_train.shape)
print('Test matrix:', X_test.shape)
print('Shared source template remains unfitted.')
```

The original source template and this implementation were fitted separately on the shared training data. Their 221 encoded feature names and all test transformation values matched exactly. See `results/audit/source_template_equivalence.json`. ADR, deposit type, customer type, special requests, season and parking remain outside the supervised contract. The encoder uses `handle_unknown="ignore"` and `min_frequency=50`; unseen categories are represented by zero columns.
